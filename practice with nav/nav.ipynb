#### Mutating and Transdforming

```python
order_enrich = order_data.merge (prod_data[['product_id', 
'unit_price']], on = 
'product_id', how ='left')
order_enrich
```

```python
order_enrich = order_enrich.assign(
    gross_rev = lambda d: d['unit_price'] * d['quantity'],
    net_rev = lambda d: d['gross_rev'] * (1 - d['discount_pct'])
    ) 

order_enrich.head()
```

### Apply

```python
def is_big_order(row):
    return((row['quantity']>=3) and (row['net_rev']>=60))
order_enrich ['big_order']= order_enrich.apply(is_big_order, axis=1) #axis=1 column wise

order_enrich.head()
```

# Summarize and aggregate

```python
order_enrich['order_ts']=pd.to_datetime(order_enrich['order_ts'])
order_enrich['order_ts'].dt.month
```

### 

```python
monthly = (order_enrich.groupby(
    [order_enrich['order_ts'].dt.month, 'channel']).agg(
        order = ('order_id', 'count'),
        qty = ('quantity', 'sum'),
        gross = ('gross_rev', 'sum'),
        net = ('net_rev', 'sum')).reset_index().rename(columns = {'order_ts' : 'month'})
)
monthly.head()
```

```python
monthly.isna().any()
```

```python
pivoted = monthly.pivot(index = 'month', columns = 'channel', values ='net')
print(pivoted)

#index=row
#column=column


pivoted_filled = pivoted.fillna(0)
print(pivoted_filled)
```

```python
long_again = pivoted_filled.reset_index().melt(id_vars='month', var_name= 'channel', value_name = 'net')

print(pivoted_filled.head())
print(long_again.sort_values('month').reset_index(drop=True).head())
```
